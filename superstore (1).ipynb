```python
# importing libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Loading Data

df = pd.read_csv(r'C:\Users\Rog_Strix\Downloads\Sample - Superstore.csv', encoding='latin1')
```

```python
# Data preview

df.head()
```

```python
df.columns
```

```python
# Bar plot for category sales


category_sales = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
```

```python
print(category_sales)
```

```python
category_sales.plot(kind='bar')
plt.title('Category wise Sales')
plt.xlabel('Category')
plt.ylabel('Sales')
plt.xticks(rotation = 45)
```

```python
# Category wise Profit

category_profit = df.groupby('Category')['Profit'].sum().sort_values(ascending=False)
print(category_profit)
```

```python
category_profit.plot(kind='bar')
plt.title('Category wise Profit')
plt.xlabel('Category')
plt.ylabel('Profit')
plt.xticks(rotation=30)
```

```python
# Region wise sales and profit


region_performance = df.groupby('Region')[['Sales','Profit']].sum().sort_values(by='Sales',ascending=False)
print(region_performance)
```

```python
region_performance.plot(kind='bar')
plt.title("Sales and Profit by Region")
plt.xlabel('Region')
plt.ylabel('Amount')
plt.xticks(rotation = 45)
```

```python
# Discount vs Profit  (Scatter Plot)

plt.scatter(df['Discount'], df['Profit'])
plt.title('Discount vs Profit')
plt.xlabel('Discount')
plt.ylabel('Profit')

```

```python
sns.scatterplot(data=df, x='Discount', y='Profit', alpha=0.5, hue='Category')
plt.title('Discount vs Profit by Category')
```

```python
avg_trend = df.groupby('Discount')[['Profit']].mean().reset_index()
```

```python
sns.lineplot(data=avg_trend, x='Discount', y='Profit')
plt.title('Discount vs Profit by Category')
```

```python
# Sales over Time


df['Order Date'] = pd.to_datetime(df['Order Date'])
```

```python
df.head()
```

```python
df['Year'] = df['Order Date'].dt.year
```

```python
df['Month'] = df['Order Date'].dt.month
```

```python
df.dtypes
```

```python
monthly_sales_y = df.groupby(['Year','Month'])['Sales'].sum()
print(monthly_sales_y)
```

```python
monthly_sales_y = monthly_sales_y.reset_index()
```

```python
monthly_sales = df.groupby(df['Order Date'].dt.month)['Sales'].sum()
print(monthly_sales)
```

```python
monthly_sales_y.plot(kind='line')

plt.title('Sales over Time')
```

```python
plt.figure(figsize=(14,6))
sns.lineplot(data = monthly_sales_y, x='Month', y = 'Sales', hue='Year')
plt.title('Monthly Sales Over Time')
plt.xticks(range(1,13))
```
